```python
# Assignment 4 
# que 1:- WAP to enhance the input image using Log Transform. Assume C=1
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('baby.jpg.jpg', 0)
img_float = np.float32(img)
log_img = np.log(1 + img_float)

log_img = (log_img / np.max(log_img)) * 255
log_img = np.uint8(log_img)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(log_img, cmap='gray')
plt.title("Log Transformed Image")
plt.axis('off')

plt.show()
```

```python
# WAP to enhance the input image using power law/gamma transform.Assume c=1, and input the value of gamma from the user.
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

gamma = float(input("Enter gamma value: "))

img_norm = img / 255.0

gamma_img = np.power(img_norm, gamma)

gamma_img = np.uint8(gamma_img * 255)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(gamma_img, cmap='gray')
plt.title(f"Gamma Corrected (γ={gamma})")
plt.axis('off')
plt.show()

cv2.imwrite("gamma_output.jpg", gamma_img)
```

```python
# que :- 3 WAP to enhance the input image using intensity level slicing (cotrast stretching). The input range of intensity values and the desired range of intensity values in the output image are obtained from the user. (example:- The input range of intensity values from 80 to 120 is stretched to 50-150 in the output image).
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

r1 = int(input("Enter lower input intensity (r1): "))
r2 = int(input("Enter upper input intensity (r2): "))

s1 = int(input("Enter lower output intensity (s1): "))
s2 = int(input("Enter upper output intensity (s2): "))

output = img.copy().astype(np.float32)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = img[i, j]

        if r1 <= r <= r2:
            output[i, j] = ((r - r1) / (r2 - r1)) * (s2 - s1) + s1
        else:
            output[i, j] = r

output = np.uint8(output)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(output, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis('off')
plt.show()

cv2.imwrite("contrast_stretched.jpg", output)
```
