```python
# 1. WAP to draw a border around the input image (or in another words pad the input image ).
# * input the width of the border (in term of pixel count ) and the desired color of the border from the user.
# * Try this for both binary and grayscale image.

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

width = int(input("Enter border width (pixels): "))
color = int(input("Enter border color (0-255): "))
bordered = cv2.copyMakeBorder(
    img,
    width,
    width,
    width,
    width,
    cv2.BORDER_CONSTANT,
    value=color
)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(bordered, cmap='gray')
plt.title("Image with Border")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

_, binary = cv2.threshold(
    img,
    127,
    255,
    cv2.THRESH_BINARY
)
width = int(input("Enter border width (pixels): "))
color = int(input("Enter border color (0 or 255): "))

bordered = cv2.copyMakeBorder(
    binary,
    width,
    width,
    width,
    width,
    cv2.BORDER_CONSTANT,
    value=color
)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(bordered, cmap='gray')
plt.title("Binary Image with Border")
plt.axis("off")
plt.show()
```

```python
# 2. WAP for image compliment try this for both binary and grayscale image.
# Program 1: Image Complement of Grayscale Image
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

complement = 255 - img

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(complement, cmap='gray')
plt.title("Complement Image")
plt.axis("off")
plt.show()
cv2.imwrite("gray_complement.jpg", complement)
```

```python
#Program 2: Image Complement of Binary Image
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baby.jpg.jpg", 0)

_, binary = cv2.threshold(
    img,
    127,
    255,
    cv2.THRESH_BINARY
)

binary_complement = 255 - binary

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(binary_complement, cmap='gray')
plt.title("Binary Complement")
plt.axis("off")
plt.show()
cv2.imwrite("binary_complement.jpg", binary_complement)
```
